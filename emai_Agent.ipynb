```ipython3
from langchain.agents import create_agent  

from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    api_key="API_KEY",
    model="gemini-2.5-flash",
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2
)
```

```ipython3
def send_email_tool(recipient: str,subject: str,body: str) ->str:
    """Sends an email to the specified recipient with given subject and body."""
    yag.send(
        to=recipient,
        subject=subject,
        contents=body
    )
    return "Email sent successfully."
```

```ipython3
graph = create_agent(
    model=llm,
    tools=[send_email_tool],
    system_prompt="You are helpfull assisstent,and you can send emails on behalf of me Asma Rao as real Email"
)
```

```ipython3
inputs = {"messages":
          [{"role":"user","content":"sent email to [RECIEVER_GMAIL] that i will be on leave tomorrow.Make it professional"}]}
```

```ipython3
for chunk in graph.stream(inputs, stream_mode="updates"):
    print(chunk)
```
